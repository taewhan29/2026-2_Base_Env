```python
def Calc_purity(clusters,Y):
    purity = 0

    for g in np.unique(clusters):
        group = Y[clusters == g]
        val, cnt = np.unique(group, return_counts = True)
        purity += np.max(cnt) / np.sum(cnt)

    return purity
```

```python
a = Calc_purity(clusters, Y)
print(a)
```
