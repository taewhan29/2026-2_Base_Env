## 기본 라이브러리 불러오기

```python
import numpy as np
import matplotlib.pyplot as plt
```

## (advance) 사용자 정의 함수로 만들기

```python
def Intersection(x,y):
    dim = x.shape[0]
    sim_val = 0 # 교집합의 개수 (둘다 1인 경우)
    for i in range(0, dim):
        if (x[i] == 1) and (y[i] == 1):
            sim_val += 1
    

    return sim_val;
```

## 데이터 벡터 x,y 생성하고, 벡터의 차원 구하기

```python
x = np.array([1,1,0,1,0,0,0])
y = np.array([1,1,1,0,1,0,0])
dim = x.shape[0]

# print(x,y,dim)
```

## for문을 이용하여, Intersection 구하기

```python
sim_val = 0 # 교집합의 개수 (둘다 1인 경우)
for i in range(0, dim):
    if (x[i] == 1) and (y[i] == 1):
        sim_val += 1

print(sim_val)
```
