## 기본 라이브러리 불러오기

```python
import numpy as np
import matplotlib.pyplot as plt
```

## (advance) 사용자 정의 함수로 만들기

```python
def jac_sim(x,y):
   (M01,M10,M00,M11) = 0,0,0,0
   dim = x.shape[0]
   for i in range(0, dim):
      if (x[i] == 0) and (y[i] == 1):
         M01 += 1
      elif (x[i] == 1) and (y[i] == 0):
         M10 += 1
      elif (x[i] == 0) and (y[i] == 0):
         M00 += 1
      elif (x[i] == 1) and (y[i] == 1):
         M11 += 1
      sim_val = (M11) / (M01 + M10 + M11)


   return sim_val;
```

## 데이터 벡터 x,y 생성하고, 벡터의 차원 구하기

```python
x = np.array([1,1,0,1,0,0,0])
y = np.array([1,1,1,0,1,0,0])
dim = x.shape[0]
```

## for문을 이용하여, 다음을 구하기

#### M01 = the number of attributes where x was 0 and y was 1
#### M10 = the number of attributes where x was 1 and y was 0
#### M00 = the number of attributes where x was 0 and y was 0
#### M11 = the number of attributes where x was 1 and y was 1

```python
(M01,M10,M00,M11) = 0,0,0,0
for i in range(0, dim):
    if (x[i] == 0) and (y[i] == 1):
       M01 += 1
    elif (x[i] == 1) and (y[i] == 0):
       M10 += 1
    elif (x[i] == 0) and (y[i] == 0):
       M00 += 1
    elif (x[i] == 1) and (y[i] == 1):
       M11 += 1


print(M01,M10,M00,M11)
```

## Jaccard Coefficient 구하기

```python
sim_val = (M11) / (M01 + M10 + M11)
print(sim_val)
```

```python
jac_sim(x,y)
```
