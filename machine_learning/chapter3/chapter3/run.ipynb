```python
from pathlib import Path

folder = Path("fr_Dataset")
files = [f.name for f in folder.glob("*.png")]

# data = np.zeros((len(files),150*150))

for i, name in enumerate(files):
    x = cv2.imread("fr_dataset/"+name)
    x = cv2.resize(x, (150,150)).flatten().astype("float")
    if i == 0:
        data = x
    else:
        data = np.vstack([data,x])
print(data)

num, dim = data.shape

# cos_sim
proxmity_mat = np.zeros((num,num))
for i in range(num):
    for j in range(num):
        x = data[i,:]
        y = data[j,:]
        if i != j:
            proxmity_mat[i,j] = mlb.cos_sim(x,y)
        else:
            proxmity_mat[i,j] = 0 # 디폴트 값

# print(proxmity_mat)

print(np.argmax(proxmity_mat,axis=1))
```
