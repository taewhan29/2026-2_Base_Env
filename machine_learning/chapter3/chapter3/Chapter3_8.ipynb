## 기본 라이브러리 불러오기

```python
import numpy as np
import matplotlib.pyplot as plt
```

## (advance) 사용자 정의 함수로 만들기

```python
def ham_dist(x,y):
    dim = x.shape[0]
    dist_val = 0
    for i in range(dim):
        if x[i] != y[i]:
            dist_val += 1

    print(dist_val)
    return dist_val;
```

## 데이터 벡터 x,y 생성하고, 벡터의 차원 구하기

```python
x = np.array([1,1,0,1,0,0,0]).astype(float)
y = np.array([1,1,1,0,1,0,0]).astype(float)
dim = x.shape[0]
```

## for문을 이용하여, L∞ norm 기반 dist_val 구하기

```python
dist_val = 0
for i in range(dim):
    if x[i] != y[i]:
        dist_val += 1

print(dist_val)
```
