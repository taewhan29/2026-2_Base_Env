## (advance) 사용자 정의 함수로 만들기

```python
def euc_distance(x,y):
    dist_val = 0
    dim = x.shape[0]
    for i in range(dim):
        dist_val = dist_val = np.abs(x[i]-y[i])**2

    dist_val = np.sqrt(dist_val)# square root
    print(dist_val)
    return dist_val;
```
