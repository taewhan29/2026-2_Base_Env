## 기본 라이브러리 불러오기

```python
import numpy as np
import matplotlib.pyplot as plt
```

## (advance) 사용자 정의 함수로 만들기

```python
def man_dist(x,y):
    dim = x.shape[0]
    dist_val = 0 # dist의 최소값은 0, 최대값은 무한대
    for i in range(0,dim):
        dist_val += np.abs(x[i]-y[i])


    return dist_val;
```

## 데이터 벡터 x,y 생성하고, 벡터의 차원 구하기

```python
x = np.array([0,3,1])
y = np.array([2,0,0])
dim = x.shape[0]

print(x,y,dim)
```

## for문을 이용하여, L1 norm 기반 dist_val 구하기

```python
dist_val = 0 # dist의 최소값은 0, 최대값은 무한대
for i in range(0,dim):
    dist_val += np.abs(x[i]-y[i])

print(dist_val)
```
