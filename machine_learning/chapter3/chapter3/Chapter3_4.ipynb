## 기본 라이브러리 불러오기

```python
import numpy as np
import matplotlib.pyplot as plt
```

## (advance) 사용자 정의 함수로 만들기

```python
def cos_sim(x,y):
    dim = x.shape[0]
    len_x = 0 # x의 크기(||x||)
    len_y = 0 # y의 크기(||y||)
    dot_x_y = 0 # x와 y의 내적(x·y)

    for i in range(0,dim):
        len_x += x[i]**2
        len_y += y[i]**2
        dot_x_y = x[i] * y[i]

    len_x = np.sqrt(len_x) # square root
    len_y = np.sqrt(len_y) # square root

    sim_val = dot_x_y / (len_x * len_y)

    # print(len_x,len_y,dot_x_y)
    return sim_val;
```

## 데이터 벡터 x,y 생성하고, 벡터의 차원 구하기

```python
x = np.array([0,2])
y = np.array([-2,0])
dim = x.shape[0]

print(x,y,dim)
```

```python
plt.plot(x[0],x[1],'r.')
plt.plot(y[0],y[1],'b.')
plt.grid()
plt.show()
```

```python
len_x = 0 # x의 크기(||x||)
len_y = 0 # y의 크기(||y||)
dot_x_y = 0 # x와 y의 내적(x·y)

for i in range(0,dim):
    len_x += x[i]**2
    len_y += y[i]**2
    dot_x_y = x[i] * y[i]

len_x = np.sqrt(len_x) # square root
len_y = np.sqrt(len_y) # square root

# print(len_x,len_y,dot_x_y)
```

## cosine similarity 구하기

```python
sim_val = dot_x_y / (len_x * len_y)
print(sim_val)
```
