```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import Normalizer
from sklearn.preprocessing import StandardScaler
```

## Data

```python
X = pd.DataFrame({
    "height" : [183,164,191,157,170],
    "weight" : [80,64,72,79,67],
    "income" : [60000, 90000, 40000, 50000, 10000]
})

num, dim = X.shape

new = pd.DataFrame({
    "height" : [170],
    "weight" : [67],
    "income" : [50000]
})
print(X)
```

## MinMaxScaler

```python
min_X = X.min(axis=0)
max_X = X.max(axis=0)
# print(min_X)
# print(max_X)

X_P = (X - min_X)/(max_X - min_X)
print(X_P)
```

```python
fig = plt.figure(figsize=(5,5))
ax = fig.add_subplot(121,projection="3d")

# plt.subplot(1,2,1)
ax.scatter(X.iloc[:,0],X.iloc[:,1],X.iloc[:,2],c='r',marker='o')
# plt.grid()

plt.subplot(1,2,2)
plt.plot(X.iloc[:,0],X.iloc[:,2],'b.')
plt.grid()
plt.show()
```

```python
result = np.sum(np.abs(np.array(X) - np.array(new)),axis=1)
print(result)
```

```python
new_P = (new - min_X) / (max_X - min_X)
result = np.sum(np.abs((np.array(X_P) - np.array(new_P))),axis=1)
print(new_P)
```

## StandardScaler

```python
mean = X.mean(axis=0)
sigma = X.std(axis=0)

X_P = (X-mean)/sigma
print(X_P)
```

```python
# val = X / np.sqrt(np.square(X).sum(axis=1, keepdims=True))
X = np.array(X)
val = X / np.sqrt(np.sum(X**2, axis=1, keepdims=True))
print(val)
```
