## 비디오 파일 불러오기

### 기본 라이브러리 불러오기

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### User parameter 정의하기

```python
delay = 30 # 영상 출력 속도 제어
```

### "Video_003.avi"를 읽어오기

### 영상 재생하기

```python
capture = cv2.VideoCapture("images/ChromaKey.mp4")
background = cv2.imread("images/back.jpg", 0)

T = 100

# 잘 읽어 왔는지 검사하는 코드
if capture.isOpened() == False: 
    raise Exception("Error")

while True:  # 무한 반복
    ret, frame = capture.read()  # 비디오 영상 순차적으로 받기
    f = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    h,w = f.shape
    background = cv2.resize(background, (w,h))
    # g = np.where(f>T, 255,0).astype("uint8")
    g = np.where(f>T,background,f).astype("uint8")

    if not ret: 
        break
        
    if cv2.waitKey(delay) >= 0:
        break
        
    cv2.imshow("video", frame)  # 윈도우에 영상 출력하기
capture.release()
cv2.destroyAllWindows()
```
