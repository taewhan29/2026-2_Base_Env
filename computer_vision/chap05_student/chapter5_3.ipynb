## Running Average

### 기본 라이브러리 불러오기

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import time
```

### 10개 프레임의 (x,y)위치값들을 정의하기

```python
f = np.array([10,12,17,13,18,20,15,111,90,14], np.uint8)
# f = np.random.randint(255, size=(1,100))
w = len(f)

print(w)
```

### running average를 활용하지 않고 평균을 업데이트하고, 속도를 구하기

```python
start = time.time()
for t in range(1,w+1):
    b = np.mean(f[0:t])
    print(str(t) + "th mean is " + str(b))

end = time.time()
print(f"{end - start:.5f} sec")
```

### running average를 활용하여 평균을 업데이트하고, 속도를 구하기

```python
start = time.time()
for t in range(1,w+1):
    if t == 1:
        b = f[t-1]
    else:
        b = (t-1)/t*b + 1/t*f[t-1]
    print(str(t) + "th mean is " + str(b))

end = time.time()
print(f"{end - start:.5f} sec")
```
