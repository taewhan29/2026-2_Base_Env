# <2차원 데이터에서의 1차 보간 : 양선형보간>

### 기본 라이브러리 불러오기

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### User parameter 정의하기

```python
# translation
tx, ty = 0, 0
# scaling
sx, sy = 2, 2
# rotation
theta = 0
theta_rad = theta * np.pi / 180
```

### 동차 행렬 생성하기

```python
# Translation matrix
T = np.array([[1,0,tx],
              [0,1,ty],
              [0,0,1]])

# Scaling matrix
S = np.array([[sx,0,0],
              [0,sy,0],
              [0,0,1]])

# Rotation matrix
R = np.array([[np.cos(theta_rad), -1*np.sin(theta_rad), 0],
              [np.sin(theta_rad), np.cos(theta_rad), 0],
              [0,0,1]])

A = S
print(A)
```

### "triangle.jpg" 흑백 영상(f)으로 불러오고,영상의 크기(h,w)를 자동으로 구하기

```python
f = cv2.imread("images/triangle.jpg",0) # gray
h,w = f.shape
```

### 결과 데이터 (g) 생성하기

```python
g = np.zeros((int(h*sx),int(w*sy)))
gh,gw = g.shape
```

### 동차 행렬, 역방향 사상을 기반으로 양선형 보간 구현하기

```python
for x_prime in range(gh):
    for y_prime in range(gw):
        P_prime = np.array([[x_prime],[y_prime],[1]])
        P = (np.linalg.inv(A)@P_prime) # 실수

         # 인접한 4개 이웃
        Q11 = (np.floor(P[:2])).astype("int16")
        Q12 = Q11 + np.array([[0],[1]])
        Q21 = Q11 + np.array([[1],[0]])
        Q22 = Q11 + np.array([[1],[1]])

        # 이웃과의 거리를 계산하기
        a = abs(P[1,0] - Q11[1,0])
        b = abs(P[0,0] - Q11[0,0])
        
        if (0 <= Q11[0,0] and Q21[0,0] < h) and (0 <= Q11[1,0] and Q12[1,0] < w):
            v1 = (1-a)*f[Q11[0,0],Q11[1,0]] + (a)*f[Q12[0,0],Q12[1,0]]
            v2 = (1-a)*f[Q21[0,0],Q21[1,0]] + (a)*f[Q22[0,0],Q22[1,0]]
            v = (1-b)*v1 + (b)*v2   
g = g.astype("uint8")
print(g)
```
