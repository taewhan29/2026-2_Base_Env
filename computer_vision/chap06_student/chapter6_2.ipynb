## <대칭(mirroring) 변환n>

### 기본 라이브러리 불러오기

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### ‘cat.jpg’를 흑백영상(f)으로 읽고, 영상의 크기(h,w)를 자동으로 구하기

```python
f = cv2.imread("images/cat.jpg",0) # gray
h,w = f.shape

print(h,w)
```

### 결과 영상(g) 생성하기

```python
g1 = np.zeros((h,w))
g2 = np.zeros((h,w))
gh, gw = g1.shape
```

### 대칭 구현하기(세로 중심축)

```python
# 미러링
M1 = np.array([[1,0,0],[0,1,w-1],[0,0,1]]) # 좌우대칭 동차좌표
M2 = np.array([[1,0,h-1],[0,1,0],[0,0,1]]) # 상하대칭
```

```python
# 𝑥^′=𝑥
# 𝑦^′=𝑤−𝑦-1 
```

```python
for x in range(0,h):
    for y in range(0,w):
        x_prime = int(x)
        y_prime = int(w-y-1)
        if (0 <= x_prime < gh) and (0 <= y_prime < gw):
            g1[x_prime,y_prime] = f[x,y]

# 8bit unsigned integer format
g1 = g1.astype("uint8")            
```

```python
### 대칭 구현하기(가로 중심축)
```

```python
# 𝑥^′=ℎ−𝑥+1
# 𝑦^′=𝑦
```

```python
for x in range(0,h):
    for y in range(0,w):
        x_prime = int(h-x+1)
        y_prime = int(y)
        if (0 <= x_prime < gh) and (0 <= y_prime < gw):
            g2[x_prime,y_prime] = f[x,y]

# 8bit unsigned integer format
g2 = g2.astype("uint8")            
```

### matplot 라이브러리를 활용하여 영상(f,g) 출력하기

```python
plt.subplot(1,3,1); plt.imshow(f, cmap="gray"); plt.axis("off")
plt.subplot(1,3,2); plt.imshow(g1, cmap="gray"); plt.axis("off")
plt.subplot(1,3,3); plt.imshow(g2, cmap="gray"); plt.axis("off")
```
