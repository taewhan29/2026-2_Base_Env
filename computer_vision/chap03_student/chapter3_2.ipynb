## 기본 라이브러리 불러오기

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

## 'Lenna.png'를 흑백영상으로 읽어오기

```python
f = cv2.imread("/workspaces/2026-2_Base_Env/computer_vision/cv_2026_docker/images/images/Lenna.png",0)
h,w = f.shape

plt.subplot(1,2,1)
plt.imshow(f, cmap='gray')
```

## 결과 영상(g)를 생성하고 반전 변환 구현하기


```python
g = cv2.imread("/workspaces/2026-2_Base_Env/computer_vision/cv_2026_docker/images/images/Lenna.png",0)

g = np.zeros((h,w))
# f[0,2]
for x in range(0,h):
    for y in range(0,w):
        g[x,y] = -f[x,y] + 255

g = np.uint8(g)
print(g)
```

```python
g = -f + 255
print(g)
```

## matplot 라이브러리를 활용하여 결과 영상 출력하기

```python
plt.subplot(1,2,1); plt.imshow(f,cmap='gray')
plt.subplot(1,2,2); plt.imshow(g,cmap='gray')
```
