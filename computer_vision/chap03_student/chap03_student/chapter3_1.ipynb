```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

## 입력값 정의하기

```python
x = 100
# print(type(x))
```

## 함수 정의하기

■ 반전 변환 함수

```python
y1 = -x + 255
print(y1)
```

■ 로그 변환 함수

```python
y2 = np.log10(x)
print(y2)
```

■ 감마 변환 함수

```python
gamma = 2
y3 = ( (x/255) ** (gamma) ) * 255
print(np.uint8(y3))
```

■ 조각 변환 함수

```python
if (x < 10):
    y4 = x ** 2
elif (x < 200):
    y4 = x / 2
else:
    y4 = x + 20
print(y4)
```

```python
def TH(x):
    if (x < 10):
        y4 = x ** 2
    elif (x < 200):
        y4 = x / 2
    else:
        y4 = x + 20    
    return y4
```

```python
y4 = TH(9)
print(y4)
```
