# <히스토그램 생성>

## 기본 라이브러리 불러오기

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

## 사용자 파라미터(parameter) 정의하기

```python
bit = 8
L = 2**bit # 3비트 이미지
num_bin = L
```

## toy example 이미지 생성하기(f)

```python
# f = np.array([[6,6,6,7],[4,5,5,3],[2,1,1,3],[0,0,1,3]])
f = cv2.imread("/workspaces/2026-2_Base_Env/computer_vision/cv_2026_docker/images/images/Lenna.png",0)
h,w = f.shape
N = h*w

print(N)
```

## 빈 히스토그램 생성하고 중첩 반복문을 활용하여 빈도수 측정하기

```python
hist_f = np.zeros(num_bin)
for x in range (0,h):
    for y in range(0,w):
        hist_f[f[x,y]] += 1
```

## matplotlib.pyplot을 이용하여 결과 히스토그램 출력하기

```python
plt.bar(np.arange(0,L))
```
