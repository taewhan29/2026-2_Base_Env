## 기본 라이브러리 불러오기

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

## 사용자 파라미터(parameter) 정의하기

```python
gamma = 1/5
```

## 'Lenna.png'를 흑백영상으로 읽어오기

```python
f = cv2.imread("/workspaces/2026-2_Base_Env/computer_vision/cv_2026_docker/images/images/image_5145865821492066621779.jpg",0)
h,w = f.shape
```

## 결과 영상(g)를 생성하고, 감마 변환 구현하기


```python
g = ((f/255) ** (gamma)) * 255
g = np.uint8(g)
print(g)
```

## 결과 영상 출력하기

```python
plt.subplot(1,2,1).imshow(f,cmap="gray")
plt.subplot(1,2,2).imshow(g,cmap="gray")
```
