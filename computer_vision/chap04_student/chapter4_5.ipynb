## Otsu 알고리즘을 활용한 이미지 분할

### 기본 라이브러리 불러오기

```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

### 히스토그램 생성 함수 정의하기

```python
def CalcHistogram(f,num_bin):
    h,w = f.shape
    hist = np.zeros(num_bin)
    
    for x in range(h):
        for y in range(w):
            hist[f[x,y]] += 1
    
    return hist;
```

### User parameter 정의하기

```python
L = 4
num_bin = L 
```

### 흑백영상(f) 생성하고, 영상의 크기(h,w)를 자동으로 구하기

```python
f = np.array([[1,2],
             [3,2]])
h,w = f.shape
N = h*w# 전체 픽셀의 갯수
```

### 정규화된 히스토그램 생성하기

```python
norm_hist = CalcHistogram(f,num_bin)/N

print(norm_hist)
```

```python
T = 3
h1 = norm_hist[0:T]
h2 = norm_hist[T:]
print(h1,h2)

w1 = np.sum(h1)
w2 = np.sum(h2)
print(w1,w2)

m1 = 0
for i in range(0,T):
    m1 = m1 + (i * norm_hist[i])
    
m1 = m1 / w1

m2 = 0
for i in range(T,L):
    m2 = m2 + (i * norm_hist[i])

m2 = m2 / w2

print(m1,m2)

v1 = 0
for i in range(0,T):
    v1 = v1 + ((i-m1)**2 * norm_hist[i])
v1 = v1 / w1

v2 = 0
for i in range(T,L):
    v2 = v2 + ((i-m2)**2 * norm_hist[i])
v2 = v2 / w2

print(v1,v2)

Sw = w1*v1 + w2*v2
print(Sw)
```

### Otsu 알고리즘 구현하기

```python
f = cv2.imread("images/seal.jpg",0)
h,w=f.shape


Sw_vec = np.zeros((L))
for threshold in range(0,L):
    w1 = np.sum(norm_hist[0:T])
    w2 = np.sum(norm_hist[T:])
    # print(w1,w2)

    m1 = 0
    for i in range(0,T):
        m1 = m1 + (i * norm_hist[i])  
    m1 = m1 / w1

    m2 = 0
    for i in range(T,L):
        m2 = m2 + (i * norm_hist[i])
    m2 = m2 / w2

    # print(m1,m2)

    v1 = 0
    for i in range(0,T):
        v1 = v1 + ((i-m1)**2 * norm_hist[i])
    v1 = v1 / w1

    v2 = 0
    for i in range(T,L):
        v2 = v2 + ((i-m2)**2 * norm_hist[i])
    v2 = v2 / w2

    # print(v1,v2)

    Sw = w1*v1 + w2*v2
    # print(Sw)

    # 각 임계값별 스코어 계산하기
Sw_vec[T] = Sw
print("T = %d, w1 = %.5f, w2 = %.5f, S1 = %.5f, S2 = %.5f, Sw = %.5f" %(threshold, w1,w2,v1,v2, Sw_vec[T]))
```

```python
plt.bar(np.arange(0,L).Sw.vec)
```

### 최적의 임계값(opt_T) 찾고 결과 이미지(g) 생성하기


```python
Sw_vec[np.isnan(Sw_vec)] = 100000
opt_T = np.argmin(Sw.vec)
print(opt_T)

g = np.where(f>opt_T
```

### matplotlib.pyplot을 이용하여 결과 이미지 출력하기


```python
plt.subplot(1,2,1);plt.imshow(f, cmap="gray"); plt.axis("off")
plt.subplot(1,2,2);plt.imshow(g, cmap="gray"); plt.axis("off")
plt.show()
```
