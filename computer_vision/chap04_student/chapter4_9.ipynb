## 지역적 임계값 처리를 적용한 세그멘테이션

### 기본 라이브러리 불러오기

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### User parameter 정의하기

```python
maxValue = 1 # 결과 영상의 최대값
thresholdType = cv2.THRESH_BINARY # 결과 저장 방식 : 이진화
blockSize = 21 # 지역의 범위
C = 5 # 상수 constant
```

### ‘text.png’를 흑백영상으로 읽고, 영상의 크기(h,w)를 자동으로 구하기

```python
f = cv2.imread() # gray
h,w = f.shape
```

### 지역적 임계값 처리용 라이브러리 함수 활용하기

```python
g_mean = cv2.adaptiveThreshold()
g_gauss = cv2.adaptiveThreshold()
```

### matplot 라이브러리를 활용하여 결과 영상 출력하기

```python
plt.subplot(3,1,1); plt.imshow(f, cmap="gray"); plt.axis("off")
plt.subplot(3,1,2); plt.imshow(g_mean, cmap="gray"); plt.axis("off")
plt.subplot(3,1,3); plt.imshow(g_gauss, cmap="gray"); plt.axis("off")
```
