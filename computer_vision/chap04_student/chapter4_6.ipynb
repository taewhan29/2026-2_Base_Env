### 사용자 정의함수 생성하기

```python
def CalcHistogram(f,num_bin):
    h,w = f.shape
    hist = np.zeros(num_bin)
    
    for x in range(h):
        for y in range(w):
            hist[f[x,y]] += 1
    
    return hist;
```
