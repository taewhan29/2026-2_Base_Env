## 히스토그램과 이미지 분할

### 기본 라이브러리 불러오기

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### 히스토그램 생성 함수 정의하기


```python
def CalcHistogram(f,num_bin):
    h,w = f.shape
    hist = np.zeros(num_bin)
    for x in range(h):
        for y in range(w):
            hist[f[x,y]] += 1
    
    return hist;
```

### User parameter 정의하기

```python
L = 256
num_bin = L
```

### ‘seal.jpg’를 흑백영상(f)으로 읽고, 영상의 크기(h,w)를 자동으로 구하기

```python
f = cv2.imread("images/seal.jpg",0) # gray
h,w = f.shape
```

### 히스토그램 생성하기(hist)

```python
hist = CalcHistogram(f,num_bin)
```

### matplot 라이브러리를 활용하여 히스토그램 출력하기

```python
plt.bar(np.arange(0,L),hist);
```

### ☆ 히스토그램의 분포를 확인하고, 적절한 임계값(threshold) 설정하기

```python
T = 140
```

### 수동 임계값 적용하여 이진 이미지(g) 만들기

```python
g = np.where(f>T,1,0)
```
