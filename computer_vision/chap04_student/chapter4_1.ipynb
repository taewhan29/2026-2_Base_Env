## 이미지 분할의 기본 개념

### 기본 라이브러리 불러오기

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### User parameter 정의하기

```python
T =  220 # 임계값(threshold)
```

### ‘seal.jpg’를 흑백영상(f)으로 읽고, 영상의 크기(h,w)를 자동으로 구하기

```python
f = cv2.imread("images/seal.jpg",0) # gray
h,w = f.shape

print(h,w)
```

### 임계값을 통해 결과 영상(g) 생성하기

#### 1번. general

```python
g = np.zeros((h,w))
for x in range(h):
    for y in range(w):
        if f[x,y] > T:
            g[x,y] = 1
        else:
            g[x,y] = 1
            
            
```

#### 2번. easy

```python
g = np.where(f>T,1,0)
```

### matplot 라이브러리를 활용하여 영상(f,g) 출력하기

```python
plt.subplot(1,2,1); plt.imshow(f,cmap='gray'); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(g,cmap='gray'); plt.axis('off')
```
