```python
x = np.zeros((2,3))
print(x)


y = np.ones((2,3))
print(y)

z = np.ones((2,3)) *100
print(z)
```

```python
x = np.array([[1,2,3],[4,5,6]])
print(np.mean(x))

print(np.mean(x,axis=1))

# 슬라이싱
print(x[:,1])
print(x[0,1])
print(x[:,:])
```

```python
A = np.array([[1,2,3],[4,5,6]])
for x in range(0,2):
    for y in range(0,3):
        A[x,y] = A[x,y]*3
        print(A)
```
